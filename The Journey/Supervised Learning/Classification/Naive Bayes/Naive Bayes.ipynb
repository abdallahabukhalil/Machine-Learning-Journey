```python
import pandas as pd
from sklearn.datasets import load_wine


wine = load_wine()

df = pd.concat(
    [
        pd.DataFrame(wine.data, columns = wine.feature_names),
        pd.DataFrame(wine.target, columns = ["wine_type"])
    ],
    axis = 1
)

df
```

```python
from sklearn.model_selection import train_test_split

train, test = train_test_split(df, test_size = 0.2, stratify = df['wine_type'], random_state = 42)
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler(feature_range = (0, 1))
scaled_train = pd.concat(
    [
        pd.DataFrame(
            scaler.fit_transform(
                train.drop('wine_type', axis = 1),
                train['wine_type']
            ),
            columns = train.drop('wine_type', axis = 1).columns,
            index = train.index
        ),
        train['wine_type']
    ],
    axis = 1
)

pd.DataFrame(scaled_train)
```

```python
from sklearn.naive_bayes import MultinomialNB

model = MultinomialNB()
model.fit(scaled_train.drop('wine_type', axis = 1), scaled_train['wine_type'])
```

```python
test_scaled = pd.concat(
    [
        pd.DataFrame(
            scaler.transform(
                test.drop('wine_type', axis = 1)
            ),
            columns =  test.drop('wine_type', axis = 1).columns,
            index =  test.drop('wine_type', axis = 1).index
        ),
        test['wine_type']
    ],
    axis = 1
)

test_scaled
```

```python
comparison = pd.concat(
    [
        pd.DataFrame(
            model.predict(
                test_scaled.drop('wine_type', axis = 1)
            ),
            columns = ['predicted_wine_type'],
        ),
        test_scaled['wine_type'].rename('actual_wine_type').reset_index(drop = True)
    ],
    axis = 1
)

comparison
```

```python
model.score(test_scaled.drop('wine_type', axis = 1), test_scaled['wine_type'])
```
